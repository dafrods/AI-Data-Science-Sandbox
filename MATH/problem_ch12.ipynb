```python
%matplotlib inline
import numpy as np
from mpl_toolkits import mplot3d
import matplotlib.pyplot as plt
```

```python

# Cycloid

psi = np.linspace(0,2*np.pi,1000)
r   = 3

x = r*(psi - np.sin(psi))
y = r*(1 - np.cos(psi))

plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Cycloid')
plt.grid()
plt.show()
```

```python
# f = lambda x : np.sqrt(((100*((1-(0.01*x**2))**2)) + 0.02*x**2)/((1-x**2)**2)+(0.1*x**2))

a = lambda x : (100*(1-(0.01*x**2))**2)+(0.02*x**2)
b = lambda x : (1-x**2)**2+(0.1*x**2)
f = lambda x : np.sqrt(a(x)/b(x))

x = np.linspace(0,100,1000)
y = f(x)


plt.subplot(2, 2, 1)
plt.plot(x,y)
plt.title('Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()

plt.subplot(2, 2, 2)
plt.semilogx(x,y)
plt.title('Semilogx')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()

plt.subplot(2, 2, 3)
plt.loglog(x,y)
plt.title('Loglog')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(which='both')

plt.subplot(2, 2, 4)
plt.semilogy(x,y)
plt.title('Semilogy')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(which='both')

plt.tight_layout()

plt.show();
```

```python

f1 = lambda x : 3 + np.exp(-x) * np.sin(6*x)
f2 = lambda x : 4 + np.exp(-x) * np.cos(6*x)

x = np.linspace(0,5,1000)

y1 = f1(x)
y2 = f2(x)


plt.plot(x,y1, 'r')
plt.plot(x,y2, 'b')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```

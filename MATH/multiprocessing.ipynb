```python
from multipr import serial_test, parallel_test
n = 1000000
```

```python
serial_out = serial_test(n)
```

```python
parallel_out = parallel_test(n)
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('seaborn-v0_8-poster')
%matplotlib inline
```

```python
n_run = np.logspace(1, 7, num = 7)

t_serial = [serial_test(int(n)) for n in n_run]
t_parallel = [parallel_test(int(n)) for n in n_run]
```

```python
plt.figure(figsize = (10, 6))
plt.plot(n_run, t_serial, '-o', label = 'serial')
plt.plot(n_run, t_parallel, '-o', label = 'parallel')
plt.loglog()
plt.legend()
plt.ylabel('Execution time (s)')
plt.xlabel('Number of random points')
plt.show()
```
